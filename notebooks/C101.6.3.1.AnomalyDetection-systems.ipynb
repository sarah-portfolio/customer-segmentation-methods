```python
def clean(data):
    characters_to_remove = [',', '$', ')', '(']  # remove financial markers
    for col in data.columns:
        if data[col].dtype == 'object':  # operate on string/object columns (R)
            for char in characters_to_remove:
                data[col] = data[col].str.replace(char, '', regex=False)
            data[col] = data[col].str.strip()
    return data

clean(data).head()
```

# The following new features are created:

```python
# Frequency is the number of orders per customer using transform to reform them
data['Frequency'] = data.groupby('Customer ID')['Order ID'].transform('count')
data['Frequency'].head()
```

```python
# Recency : N.B. Note that Order Date is a new column and NOT 'Order_Date'
from datetime import datetime
from dateutil import parser # amenable to multiple non-specified date formats

# Apply a custom function to parse the dates
def parse_date(date_str):
  return parser.parse(date_str)

data['Delivery Date'] = data['Delivery_Date'].apply(parse_date)
current_date = pd.to_datetime(datetime.now())
data['Delivery Date'] = pd.to_datetime(data['Delivery_Date'])

# Calculation of days to delivery date by using days from Delivery _ and current_.
data['Recency'] = (current_date - data['Delivery Date']).dt.days
data['Recency'].head()
```

```python
# Average Unit Costs

data['Unit Cost'] = pd.to_numeric(data['Unit Cost'], errors='coerce') # Convert 'Unit Cost' to numeric, handling errors


# Mean of AUC because there are multiple orders and unit cost entries
data['Average Unit Cost'] = data.groupby('Customer ID')['Unit Cost'].transform('mean')


#data['Average Unit Cost
data['Average Unit Cost'].head()
#data['Average Unit Cost'].describe()
```

CLV=Average Order Value×Purchase Frequency×Customer Lifespan

Average Order Value (AOV) = Total Revenue / Number of Orders

Purchase Frequency = Number of Orders / Number of Customers

Customer Lifespan = Average length of customer relationship (days/months/years)

```python
# Customer Lifetime Value
data['Total Revenue'] = data['Total Revenue'].astype(float)
data['Total Revenue'].head()
```

```python
data['CLV'] = data.groupby('Customer ID')['Total Revenue'].transform('sum')
data['CLV'].describe()
```

```python
# Customer Age ((explore format='ISO8601'))
data['DOB'] = pd.to_datetime(data['Customer_BirthDate'].apply(parse_date))
data['DOB'].head()
```

```python
data['Age'] = (current_date - data['DOB']).dt.days // 365
data['Age'].head()
```

```python
# Average Order Value (AOV) = Total Revenue / Number of Orders
data['Average Order Value'] = data['Total Revenue'] / data['Frequency']
data['Average Order Value'].head()
```

# Performed hierarchical clustering and visualised the dendrogram

```python
# Produce a Dendogram https://scikit-learn.org/stable/auto_examples/cluster/plot_agglomerative_dendrogram.html#sphx-glr-auto-examples-cluster-plot-agglomerative-dendrogram-py

from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

# Use the scaled numerical data
X = X_scaled

# Step 1: Fit AgglomerativeClustering with n_clusters
model = AgglomerativeClustering(distance_threshold=None, n_clusters=6)
model = model.fit(X)

# Compute the linkage matrix for the dendrogram using scipy
# Since sklearn's AgglomerativeClustering does not expose distances, we recreate it here
linkage_matrix = linkage(X, method='ward', metric='euclidean')

# Plotting the dendrogram
def plot_dendrogram(model, linkage_matrix, **kwargs):
    """
    Plots a dendrogram using a precomputed linkage matrix.

    Parameters:
    - model: the fitted AgglomerativeClustering model
    - linkage_matrix: result from scipy.cluster.hierarchy.linkage
    - **kwargs: extra keyword arguments passed to the dendrogram function (e.g., truncate_mode, p)
    """
    # Create counts array: number of original samples under each merge step
    counts = np.zeros(model.children_.shape[0])
    n_samples = len(model.labels_)

    for i, merge in enumerate(model.children_):
        current_count = 0
        for child_idx in merge:
            # child_idx refers to an index in the linkage tree
            if child_idx < n_samples:
                # If index is less than number of samples, it's a leaf node (original point)
                current_count += 1
            else:
                # Else, it's a node that's already been merged earlier
                current_count += counts[child_idx - n_samples]
        counts[i] = current_count

    # Plot dendrogram using the provided linkage matrix
    dendrogram(linkage_matrix, **kwargs)

# Show the dendrogram
plt.figure(figsize=(22, 10))
plt.title('Hierarchical Clustering Dendrogram of the Full Dataset')

# Truncate the tree to the top p levels to simplify the view
plot_dendrogram(model, linkage_matrix=linkage_matrix, truncate_mode="level", p=7)

plt.xlabel("Data point index or size of merged cluster")
plt.ylabel("Distance")

# Rotate x-axis labels to avoid overlap
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()

# Dendogram clearly shows the simplicity of the split at k=4, and the rationalse for k=6, which one branch clearly forms into, and also k=7, which is the point where the second cluster begins to split at almost the precise point that another node has split.
```
